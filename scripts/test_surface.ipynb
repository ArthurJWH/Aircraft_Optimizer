```julia
function new_plane()
    plain = Airfoil("../assets/airfoils/Plain/Plain.dat")
    wing = Aerosurface(
        airfoils=[plain, plain],
        b=6.0,
        chord=y -> 2 * (1 - y^2)^0.5,
        sw_center=0.5,
        mirror_xz=false
    )
    wing2 = Aerosurface(
        airfoils=[plain, plain],
        b=6.0,
        chord=y -> 2 * (1 - (1-y)^2)^0.5,
        sw_center=0.5,
        mirror_xz=false,
        pos=(1.0, -6.0, 0.0)
    )
    return Plane([wing, wing2])
end
```

```julia
plane = new_plane()
```

```julia
mesh = VLMMesh(plane, 32, 32)
```

```julia
p = plot_mesh(mesh)
```
